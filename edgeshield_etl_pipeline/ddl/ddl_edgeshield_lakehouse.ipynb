# EdgeShield Lakehouse setup (DLT-first)

**Required:** create catalog + bronze/silver/gold **schemas** only.

**Do not** pre-create silver or gold tables.  
DLT creates them from `transformations/bronze|silver|gold`.

```python
%sql
CREATE CATALOG IF NOT EXISTS edgeshield_lakehouse;
USE CATALOG edgeshield_lakehouse;

CREATE SCHEMA IF NOT EXISTS bronze;
CREATE SCHEMA IF NOT EXISTS silver;
CREATE SCHEMA IF NOT EXISTS gold;
```
