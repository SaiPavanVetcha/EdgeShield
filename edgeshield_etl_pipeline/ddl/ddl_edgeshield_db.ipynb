```python
%sql
CREATE DATABASE IF NOT EXISTS edgeshield_db;
USE edgeshield_db;
```

```python
%sql
-- Landing tables (server writes here). room is on system_metrics only.

CREATE TABLE IF NOT EXISTS edgeshield_db.system_metrics (
    event_id STRING NOT NULL,
    device_id STRING NOT NULL,
    hostname STRING,
    room STRING,
    timestamp TIMESTAMP NOT NULL,
    platform STRING,
    boot_time_utc TIMESTAMP
) USING DELTA;

CREATE TABLE IF NOT EXISTS edgeshield_db.cpu_metrics (
    event_id STRING NOT NULL,
    device_id STRING NOT NULL,
    timestamp TIMESTAMP NOT NULL,
    utilization_percent DOUBLE,
    logical_cores INT,
    physical_cores INT,
    frequency_current_mhz DOUBLE,
    frequency_max_mhz DOUBLE
) USING DELTA;

CREATE TABLE IF NOT EXISTS edgeshield_db.ram_metrics (
    event_id STRING NOT NULL,
    device_id STRING NOT NULL,
    timestamp TIMESTAMP NOT NULL,
    total_bytes BIGINT,
    used_bytes BIGINT,
    available_bytes BIGINT,
    utilization_percent DOUBLE,
    available_mb DOUBLE
) USING DELTA;

CREATE TABLE IF NOT EXISTS edgeshield_db.disk_metrics (
    event_id STRING NOT NULL,
    device_id STRING NOT NULL,
    timestamp TIMESTAMP NOT NULL,
    mount_point STRING,
    total_bytes BIGINT,
    used_bytes BIGINT,
    free_bytes BIGINT,
    free_gb DOUBLE,
    utilization_percent DOUBLE,
    read_bytes BIGINT,
    write_bytes BIGINT
) USING DELTA;

CREATE TABLE IF NOT EXISTS edgeshield_db.battery_metrics (
    event_id STRING NOT NULL,
    device_id STRING NOT NULL,
    timestamp TIMESTAMP NOT NULL,
    present BOOLEAN,
    percent DOUBLE,
    power_plugged BOOLEAN,
    design_capacity_raw BIGINT,
    full_charge_capacity_raw BIGINT,
    health_percentage DOUBLE
) USING DELTA;

CREATE TABLE IF NOT EXISTS edgeshield_db.environment_metrics (
    event_id STRING NOT NULL,
    device_id STRING NOT NULL,
    timestamp TIMESTAMP NOT NULL,
    temperature_c DOUBLE,
    humidity_percent DOUBLE
) USING DELTA;
```
